# Daytona sandbox setup for the fintech project

This notebook verifies the project structure, dependency readiness, and Daytona configuration before creating a sandbox for the fintech app.

## Outline
1. Inspect project structure
2. Verify Python and Node.js dependencies
3. Check Daytona SDK installation
4. Validate environment variables and API key
5. Initialize the Daytona client and configuration
6. Create the sandbox and run a Hello World script
7. Configure backend/frontend runtime details and preview URLs
8. Verify PostgreSQL and PayPal sandbox readiness
9. Capture final status and remaining requirements

```python
import os
import importlib.util
from pathlib import Path

root = Path('d:/newassign')
backend = root / 'fintech-chatbot' / 'backend'
frontend = root / 'fintech-chatbot' / 'frontend'

print('project_root_exists=', root.exists())
print('backend_exists=', backend.exists())
print('frontend_exists=', frontend.exists())
print('daytona_sdk_installed=', importlib.util.find_spec('daytona') is not None)
print('daytona_api_key_present=', bool(os.getenv('DAYTONA_API_KEY')))

if not os.getenv('DAYTONA_API_KEY'):
    print('Set the DAYTONA_API_KEY environment variable before creating a sandbox.')
    print('Need a key? Visit: https://app.daytona.io/dashboard/keys')
else:
    try:
        from daytona import Daytona, DaytonaConfig
        config = DaytonaConfig(api_key=os.environ['DAYTONA_API_KEY'])
        client = Daytona(config)
        sandbox = client.create()
        print('sandbox_created=', bool(sandbox))
        response = sandbox.process.code_run('print("Hello World from Daytona sandbox")')
        print('exit_code=', response.exit_code)
        print('result=', response.result)
        sandbox.delete()
        print('sandbox_deleted=', True)
    except Exception as exc:
        print(type(exc).__name__)
        print(str(exc))
        print('Most common fix: set a default region in the Daytona Dashboard for this organization.')
```
